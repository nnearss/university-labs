# 3. Коллекции и функция анализа продаж

В пособии приведён пример расчёта выручки, числа чеков и уникальных цен. По условию дополнительного задания код необходимо изменить так, чтобы он рассчитывал ещё и **средний чек**.


```python
def analyze_sales(price_list):
    total_revenue = sum(price_list)
    orders_count = len(price_list)
    unique_prices = set(price_list)
    average_check = total_revenue / orders_count if orders_count else 0
    return total_revenue, orders_count, unique_prices, average_check

sales = [450, 1200, 450, 2300, 1200]
revenue, orders_count, unique_prices, average_check = analyze_sales(sales)

print("--- Отчет за час ---")
print(f"Выручка: {revenue} руб.")
print(f"Заказов: {orders_count} шт.")
print(f"Уникальные цены в чеках: {sorted(unique_prices)}")
print(f"Средний чек: {average_check:.2f} руб.")
```

# 4. Самостоятельная работа 1  
## Задача «Анализ активности зрителя»

Исходная строка содержит идентификатор пользователя и список включённых фильмов, среди которых есть повторения.

Необходимо:

1. очистить строку от пробелов и убрать префикс с ID;
2. разделить строку на названия фильмов;
3. определить уникальные фильмы;
4. собрать кортеж с ID пользователя, общим числом просмотров и числом уникальных названий;
5. вывести итоговый отчёт.

Вместо жёсткого номера среза используется позиция двоеточия. Это сохраняет требуемое использование среза и делает код устойчивым, если длина ID изменится.


```python
raw_data = " ЗРИТЕЛЬ_505:начало;интерстеллар;начало;довод;начало "

# 1. Очистка строки и выделение части после префикса
prepared_string = raw_data.strip()
prefix_end = prepared_string.index(":") + 1
clean_string = prepared_string[prefix_end:]

# 2. Превращение строки в список фильмов
movies_list = clean_string.split(";")

# 3. Выделение уникальных названий
unique_movies = set(movies_list)

# 4. Сбор итоговых данных в кортеж
user_id = prepared_string.split("_")[1].split(":")[0]
user_report = (user_id, len(movies_list), len(unique_movies))

# 5. Итоговый отчет
print("--- ОТЧЕТ ОБ АКТИВНОСТИ ЗРИТЕЛЯ ---")
print(f"Пользователь: {user_report[0]}")
print(f"Всего просмотров: {user_report[1]}")
print(f"Уникальных фильмов: {user_report[2]}")
print(f"Список просмотров: {movies_list}")
print(f"Уникальные названия: {sorted(unique_movies)}")

assert user_report == ("505", 5, 3)
```

### Вывод по самостоятельной работе 1

Пользователь включал фильмы пять раз, но уникальных фильмов оказалось только три. Задача показывает, как с помощью строковых методов и коллекций можно преобразовать неструктурированную лог-запись в краткий аналитический отчёт.

## Комплексные задачи

Комплексные задачи объединяют сразу несколько изученных инструментов: строки, срезы, списки, циклы, операторы `continue` и `break`, словари, множества, функции и кортежи.


## Комплексная задача 1. Очистка данных и поиск аномалий

В лог-строке встречается запись об ошибке и аномально высокая цена.  
Согласно условию:

- элемент `ERROR` необходимо пропустить;
- при цене более 50 000 рублей цикл необходимо остановить;
- по обработанным корректным значениям рассчитывается средняя цена.


```python
data = "PRICE:450;PRICE:1200;ERROR:0;PRICE:2300;PRICE:99999;PRICE:1500"

items = data.strip().split(";")
clean_prices = []
anomaly = None

for item in items:
    if item.startswith("ERROR"):
        continue

    price_value = int(item.split(":")[1])

    if price_value > 50000:
        anomaly = price_value
        break

    clean_prices.append(price_value)

average_price = sum(clean_prices) / len(clean_prices) if clean_prices else 0

print("--- ОЧИСТКА ЦЕНОВЫХ ДАННЫХ ---")
print(f"Корректные цены до обнаружения аномалии: {clean_prices}")
print(f"Обнаруженная аномалия: {anomaly} руб.")
print(f"Средняя цена по корректным данным: {average_price:.2f} руб.")

assert clean_prices == [450, 1200, 2300]
assert anomaly == 99999
```

### Интерпретация результата

Значение `ERROR:0` не попало в расчёт, поскольку является ошибочной записью. После обнаружения аномальной цены 99 999 рублей цикл остановлен, поэтому последующее значение 1500 рублей намеренно не включено в итоговый список в соответствии с условием задачи.


## Комплексная задача 2. Сведение отчёта из разных источников


```python
vegetables = ["огурцы", "помидоры", "огурцы", "лук", "помидоры"]
amounts = [1.5, 2, 0.5, 1, 1]

unique_vegetables = set(vegetables)
final_report = {}

for vegetable, amount in zip(vegetables, amounts):
    if vegetable in final_report:
        final_report[vegetable] += amount
    else:
        final_report[vegetable] = amount

print("--- ИТОГОВЫЙ ОТЧЕТ ПО ПОКУПКАМ ---")
for vegetable, total_amount in final_report.items():
    print(f"{vegetable.capitalize()}: {total_amount:g} кг")

print(f"Количество уникальных видов овощей: {len(unique_vegetables)}")
print(f"Словарь итоговых данных: {final_report}")

assert final_report == {"огурцы": 2.0, "помидоры": 3, "лук": 1}
```

### Интерпретация результата

Два исходных списка сведены в один словарь. Повторные записи по одинаковым овощам объединены: приобретено 2 кг огурцов, 3 кг помидоров и 1 кг лука. Множество подтверждает, что в покупке присутствуют три уникальные позиции.


## Комплексная задача 3. Упаковка результатов модели


```python
def evaluate_model(predictions, targets):
    model_name = "Базовая модель классификации"
    total_tests = len(predictions)
    matches = sum(
        prediction == target
        for prediction, target in zip(predictions, targets)
    )
    accuracy = matches / total_tests if total_tests else 0
    return model_name, accuracy, total_tests

preds = [1, 0, 1, 1]
targs = [1, 0, 0, 1]

name, accuracy, tests_count = evaluate_model(preds, targs)

print(
    f"Модель {name} показала точность {accuracy:.1%} "
    f"на тесте из {tests_count} примеров."
)

assert accuracy == 0.75
assert tests_count == 4
```

### Интерпретация результата

Из четырёх предсказаний модели три совпали с фактическими ответами. Поэтому точность составила 75%. Функция возвращает результат в виде кортежа, что удобно для последующей распаковки и оформления отчёта.

# Ответы на контрольные вопросы

**1. Для чего используется метод `.strip()`?**  
Метод удаляет пробелы и другие служебные символы по краям строки. Это необходимо перед обработкой данных, чтобы одинаковые значения не воспринимались программой как разные.

**2. Чем список отличается от множества?**  
Список сохраняет порядок элементов и допускает повторения. Множество хранит только уникальные элементы и удобно для подсчёта уникальных значений.

**3. Для чего применяется условие `if`?**  
Условие позволяет выполнять разные действия в зависимости от значения данных, например пропускать ошибочные записи или проверять корректность пароля.

**4. Чем отличаются `continue` и `break`?**  
`continue` пропускает только текущую итерацию цикла и продолжает обработку. `break` полностью прекращает выполнение цикла.

**5. Зачем нужны функции?**  
Функции позволяют один раз описать повторяемую операцию и затем применять её к разным данным. Это снижает объём кода и уменьшает вероятность ошибок.
